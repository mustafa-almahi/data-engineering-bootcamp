```python
import pandas as pd
df = pd.read_csv("sales_messy.csv")

display(df["date"])


# display(df.head())
# display(df.tail())
# print(df.describe())



print(df.isnull().sum())
print(df.dtypes)
```

```python

# Applying the function to remove extra spaces and capitalizing the first letter

df["product"]= df["product"].str.strip().str.title()
#display(df["product"])

df["customer_name"]= df["customer_name"].str.strip().str.title()
#display(df["customer_name"])



#print(df["product"].unique())
Unique_products = df["product"].unique() 
print(Unique_products)

```

```python
# Handling amount column  - Filling the missing values

# print(df["amount"].describe())
# print(df["amount"].count())

df["amount"] = df["amount"].fillna(df["amount"].mean())

display(df["amount"])
display(df)

# Fixing the customer name column
df["customer_name"]= df["customer_name"].str.strip().str.title()


#===================================================================================
# df["amount"] = df["amount"].str.strip().str.replace("$","").astype(float)
# display(df["amount"])
# df["date"] = pd.to_datetime(df["date"])
# display(df["date")
```

```python
# Fix  dates column

#display(df["date"])
df['date']=pd.to_datetime(df["date"], errors ="coerce")
display(df["date"])

#Fill the Null date
df['date']= df["date"].fillna(method='ffill')
display(df["date"])



```

```python
# Add a new column called 'year' and 'month' to the dataframe

df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
display(df)


df.to_csv("salse_cleaned.csv", index=False)
df.to_csv("salse_cleaned_index.csv", index=True)


```
