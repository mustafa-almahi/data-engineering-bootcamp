```python
import pandas as pd

df =pd.read_csv('employees.csv')
display(df)
```

```python

display(df)

print(df.head(3))
print(df.describe())
print(df.tail)
print(df.columns)
print(df.dtypes)


print(df.shape)  # 'Shape Will Print (num row , num column )
print(df.info())

# Checking Nulls
print(df.isnull())
print(df.notnull())
print(df.isnull().sum())
```

```python


# Drop all rows With Null Values from Specific Subset
df_modified = df.dropna(subset=['salary','email']) 
display(df_modified)


# Fill Nulls With Defualt Value 

df['department'] = df['department'].fillna('Unknown')
display(df)
df['salary'] = df['salary'].fillna(df['salary'].median())
display(df)


# Convert and clean date column
df['join_date'] = pd.to_datetime(df['join_date'], errors='coerce')
display(df)
df['join_date'] = df['join_date'].dt.strftime('%d-%m-%Y')
display(df)


# Filtter CSV File 
higher_salary = df[df['salary'] > 7000]
display(higher_salary)




### WHY use dropna() Vs fillna()?
# Dropna() is used when missing data makes the row unusable. For example, if you have a dataset of customer 
# Fillna() is used when reasonable default can be applieyd
 

### WHY use dropna() instead of drop()?
# The dropna() method is used to drop rows or columns that contain missing values (NaN) from a pandas DataFrame or Series. It is a convenient way to remove rows or columns that have missing data, which can be useful for data cleaning and preprocessing.



```
