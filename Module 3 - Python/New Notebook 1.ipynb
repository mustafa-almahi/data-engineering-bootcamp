## ****************** Series *****************


```python
salaries = pd.Series([1000,2000,5000,10000])
print(salaries)
display(salaries)
type(salaries)
```

### Creating Custom Index


```python
# allowes named  named lookup and simulates dictionary-like behavior

employees = pd.Series([1000,2000,5000,10000], index=['John','Mary','Bob','Jane'])
print(employees)
print(employees["Bob"])

```

###Sereis with mixed data type


```python
# Sereis with mixed type
details = pd.Series(['John', 30,'London', 'Engineer'])
print(details)
print(details[2])
```

```python
# Apply Math on Series
salaries = pd.Series([1500,2000,3700,6000])

salaries_hike=salaries * 1.10
print(salaries_hike)


# Apply function on Series  

print(salaries)
print(salaries+1000)
print(salaries*1.05)
print(salaries/1000)
print(salaries**2)
print(salaries.max())
print(salaries.min())
print(salaries.mean())
print(salaries.median())
print(salaries.std())
print(salaries.describe())

print(salaries.sort_values)
print(salaries.sort_index)

print(salaries.sort_values(ascending=False))
print(salaries.sort_index(ascending=True))
print(salaries.sort_index(ascending=False))

print(salaries.sort_values(ascending=False).head(2))
print(salaries.sort_index(ascending=False).head(2))

updated_salaries=salaries.apply(lambda x: x*1.30)
print(updated_salaries)
```

```python
# Handling NUlls
data=pd.Series([1000,None,3000,4000])
print('Orignal Series\n',data)
print('Series with nulls dropped\n',data.dropna())

print(data.dropna())
print(data.isnull())
print(data.notnull())

print('********************************************************************************************')

print('Series with nulls filled\n',data.fillna(5))
print('Series with nulls filled with mean\n',data.fillna(data.mean()))
print('Series with nulls filled with median\n',data.fillna(data.median()))
print('Series with nulls filled with mode\n',data.fillna(data.mode()))
print('Series with nulls filled with mode\n',data)
```

## ****************** Data Frames *****************


```python
# A Pandas DataFrame is a two-dimensional, tabular data structure with labeled axes (rows and columns) used for data manipulation and analysis in Python

# A Pandas DataFrame resides entirely in your computer's Random Access Memory (RAM) because it is engineered for maximum execution speed. This architectural design choice ensures that every calculation, filter, and transformation executes nearly instantaneously, rather than waiting on slow hard drive read/write cycles.

```

```python

df = pd.DataFrame({'Name': ['John', 'Mary', 'Bob', 'Jane'],
                   'Age': [30, 25, 40, 35],
                   'City': ['London', 'Paris', 'London', 'Paris'],
                   'Salary': [1000, 4000, 6500,3000]
                   })


display(df)

print(df)
print(df.head())
print(df.tail())
print(df.describe())
print(df.info())
print(df.shape)
print(df.columns)
print(df.index)
print(df.dtypes)
print(df.isnull())
print(df.notnull())
print(df.isnull().sum())
print(df.isnull().sum().sum())
print(df.isnull().sum().sum()/df.size)
```

```python
Orders={'OrderID':[1001,1002,1003,1004,1005],
            'CustomerID':[101,102,103,104,105],
            'Product':['Laptop','Phone','Tablet','Monitor','Keyboard'],
            'Quantity':[1,2,1,3,2],
            'Price':[1000,500,300,200,100]}



df_01 =pd.DataFrame(Orders)
#display(pd.DataFrame(Orders))
display(df_01)

Customers={'CustomerID':[101,102,103,104,105],
        'Name':['John','Mary','Bob','Jane','Emma'],
        'City':['London','Paris','London','Paris','Berlin'],
        'Country':['UK','France','UK','France','Germany']}

df_02 =pd.DataFrame(Customers)
#display(pd.DataFrame(Customers))
display(df_02)



# Merge DataFrames
#df_03 = pd.merge(df_01,df_02, on='CustomerID')
df_merge = pd.merge(df_01,df_02, on='CustomerID')
display(df_merge)




```

```python
display(df_merge)



# Column Access --------------------

display(Customers['Name'])
display(Orders['Product'])
display(df_merge["City"])


# =================================================================================

# Row level Access --------------------

print(df_01.loc[0])    # Index label        -->   Row with index label 0
print(df_01.iloc[0])   # Row position       -->   First row  

print(df_01.loc[[0,1]])    # Index label    -->   Rows with index labels 0 and 1
print(df_01.iloc[[0,1]])   # Row position   -->   First and second rows

# Use .loc[] when you know the index labels.
# Use .iloc[] when you want rows or columns by their position (first, second, third, etc.)

# =================================================================================

# Value level Access --------------------

print(df_01.loc[0, "Product"])         #  index label is 0
print(df_02.loc[1, "Name"])            #  index label is 0
print(df_02.loc[2, "City"])            #  index label is 2


```

```python


# Adding new columns --------------------

df_merge['total']= df_merge['Quantity'] * df_merge['Price']
display(df_merge)


# =================================================================================

# Filtering --------------------

print(df_01.loc[df_01['Quantity'] > 1])
print(df_01.loc[df_01['Quantity'] > 1, 'Product'])

print(df_01.loc[df_01['Quantity'] > 1, ['Product','Price']])
print(df_01.loc[df_01['Quantity'] > 1, ['Product','Price']].reset_index(drop=True))
print(df_01.loc[df_01['Quantity'] > 1, ['Product','Price']].reset_index(drop=True).sort_values(by='Price', ascending=False))
```
