```python
%sql
CREATE OR REPLACE TABLE employees AS
SELECT DISTINCT *
FROM employees;
```

```python
%sql
select * from employees
```

```python
%sql    
select first_name ,last_name , job_title , salary 
from employees
order by salary desc limit 10 offset 3 -- skip first 3 rows
        
```

```python
%sql
select * from employees
where department = 'Sales'
```

```python
%sql
select * from employees
where department = 'Sales'
```

```python
%sql
select * from employees
where city in ('Hyderabad', 'Bangalore')  and hire_date > '2020-01-01'
    
--select * from employees
```

```python
%sql
select  department , sum(salary) as total_salary
from employees
group by department
having total_salary > 1000000.00
```

```python
%sql
select * from employees
where department = 'Sales'
order by salary desc
```

```python
%sql
select * from employees order by department asc, salary desc  , first_name asc
--select * from employees order by department asc, first_name asc, salary desc
```

```python
%sql    
select first_name ,hire_date , job_title  
from employees
order by hire_date  desc limit 5 offset 2 -- skip first 2 rows
```

```python
%sql
select emp_id, first_name , job_title, salary from employees
order by salary desc limit 5 offset 1 -- second highest salary
```

```python
%sql
--select distinct department from employees
select distinct city  from employees
```

```python
%sql
-- finding unique years
select distinct year(hire_date)
from employees
```

```python
%sql
-- finding how many employees were hired in each year
select distinct year(hire_date) as YRS, count(hire_date) as total_hires
from employees
group by YRS --year(hire_date)
```
