```sql
%sql
--   *** Subqueries ***   
-- query inside query "nested query"   used for filtering data 

--  Find customer who placed orders above the average order amount  #1    

select distinct c.first_name , c.last_name 
from customers c 
join orders o 
on c.customer_id = o.customer_id where total_amount >
(
select AVG(total_amount) from orders
)
order by c.first_name


```

```sql
%sql
--  Find Customer who placed orders above the average order amount #2     

select distinct c.first_name , c.last_name , MAX(o.total_amount) as highest_sale
from customers c 
join orders o 
on c.customer_id = o.customer_id where total_amount >
(
select AVG(total_amount) from orders
)
group by c.first_name , c.last_name
order by highest_sale
```

```sql
-- List all products have been sold more than 2 times in total.

select od.product_name , sum(od.quantity)
from order_details od 
group by od.product_name
having sum(od.quantity) > 2     -- (select 2)
order by sum(od.quantity) desc
```

```sql
-- LIST Products WITH thier total REVENUE

WITH PRODUCT_REVENUE AS  (
  select  
  od.product_name as product_name , 
  count(od.quantity) as total_quantity ,
  sum(od.quantity * od.price_each) as total_revenue
  from  order_details od
  group by od.product_name
)

select * from PRODUCT_REVENUE
```

```sql
-- *** Case statement ***
-- used to add additonal logic to qurey 
-- Usefull fro lable data or categorize data based on certain conditions.and handel exception

-- Categrize oreders based on thier total amount as SMALL , MEDIUM and LARGE

select o.order_id , o.total_amount,
  CASE 
    WHEN o.total_amount<1000 THEN "SMALL"
    WHEN o.total_amount between  1000 and 2500 THEN "MEDIUM"
    ELSE "LARGE"
  END AS order_size
from orders o
order by total_amount desc
```

```sql
-- Lable CUSTOMER as "New" or "Returning" based on order Count

Select   c.first_name , count(o.order_id) total_orders ,
  CASE 
    WHEN COUNT(o.order_id) = 1 THEN "New"
    WHEN COUNT(o.order_id) > 1 THEN "Returning"
    ELSE "No Orders"
  END AS customer_status
From customers c left join orders o on c.customer_id = o.customer_id
group by c.first_name


```

```sql
--  "WITH CTE" Lable CUSTOMER as "New" or "Returning" based on order Count

WITH CUSTOMER_lABLE AS(
Select   c.first_name , count(o.order_id) total_orders ,
  CASE 
    WHEN COUNT(o.order_id) = 1 THEN "New"
    WHEN COUNT(o.order_id) > 1 THEN "Returning"
    ELSE "No Orders"
  END AS customer_status
From customers c left join orders o on c.customer_id = o.customer_id
group by c.first_name
)

SELECT * FROM CUSTOMER_lABLE
WHERE customer_status = "Returning"
```

```sql
-- Count Orders for each month #1
SELECT 
  DATE_FORMAT(o.order_date, 'MMMM') AS month,
  COUNT(o.order_id) AS total_orders
FROM orders o 
GROUP BY month
```

```sql
-- Count Orders for each month #2
SELECT 
  MONTH(o.order_date) AS order_month, COUNT(o.order_id) AS total_orders
FROM orders o
GROUP BY order_month
ORDER BY order_month 

/*  
SELECT 
  MONTH(o.order_date) AS order_month, COUNT(o.order_id) AS total_orders
FROM orders o
GROUP BY 1
ORDER BY 1
*/
```

```sql
-- LIST ALL Customer's Name + Spend + Category

WITH customer_spend AS (
select   c.customer_id , concat (c.first_name , " " , c.last_name) as customer_name , sum(od.quantity  * od.price_each) as Total_spend
from customers c 
inner join  orders o  on c.customer_id = o.customer_id
inner join order_details od   on o.order_id = od.order_id
group by c.customer_id , c.first_name , c.last_name
order by total_spend desc
)

SELECT customer_name , Total_spend , 
  CASE 
    WHEN Total_spend < 15000 then "silver"
    WHEN Total_spend between 15000 and 30000 then "Bronze"
  ELSE "Gold"
  END AS customer_category
FROM customer_spend





```
