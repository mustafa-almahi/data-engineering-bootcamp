```sql
%sql
--  Find customer who placed orders above the average order amount  #1    

select distinct c.first_name , c.last_name 
from customers c 
join orders o 
on c.customer_id = o.customer_id where total_amount >
(
select AVG(total_amount) from orders
)
order by c.first_name


```

```sql
%sql
--  Find Customer who placed orders above the average order amount #2     

select distinct c.first_name , c.last_name , MAX(o.total_amount) as highest_sale
from customers c 
join orders o 
on c.customer_id = o.customer_id where total_amount >
(
select AVG(total_amount) from orders
)
group by c.first_name , c.last_name
order by highest_sale
```

```sql
-- List all products have been sold more than 2 times in total.

select od.product_name , sum(od.quantity)
from order_details od 
group by od.product_name
having sum(od.quantity) > 2     -- (select 2)
order by sum(od.quantity) desc
```
