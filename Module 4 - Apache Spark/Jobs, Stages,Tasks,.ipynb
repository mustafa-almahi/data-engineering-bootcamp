```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Data Reading
df = spark.read.format("csv")\
            .option("header",True)\
            .option("infarSchema",True)\
            .load("/Workspace/Users/m.osman.almahi@gmail.com/data-engineering-bootcamp/Module 4 - Apache Spark/MegaMart.csv")

# Transformations
df = df.filter(col('product_name')=="Power Bank")
df = df.select('order_id','product_name')
df = df.groupBy('product_name').agg(count(col('order_id')))

#Action
display(df)
```
