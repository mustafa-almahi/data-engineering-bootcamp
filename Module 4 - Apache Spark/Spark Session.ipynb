```python
data = [
    ("Alice", 25, "New York"),
    ("Bob", 30, "San Francisco"),
    ("Charlie", 35, "Chicago")
]

schema = StructType([
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("city", StringType(), True)
])


df = spark.createDataFrame(data, schema=schema)
```

```python
# display(data)
df = spark.createDataFrame(data, schema)
display(df)
```
